```python
import matplotlib.pyplot as plt
import seaborn as sns
import pandas as pd
```

```python
# Set the Seaborn style
sns.set_theme(style="white")

# Prepare the data
models = ['ScaleSense v1 YOLOv5', 'ScaleSense v2 YOLOv5', 'ScaleSense v2 YOLOv7']
values = [128, 154, 167]

# Create a dataframe
data = pd.DataFrame(list(zip(models, values)), columns=['Метод', 'Кількість опрацьованих зображень'])

# Create a list of colors
# 'hotpink' for gummy pink, 'lightcyan' for something like tiffany and 'violet' for violet.
colors = ['hotpink', 'turquoise', 'indigo']

# Create the plot
plt.figure(figsize=(10, 6))
barplot = sns.barplot(x='Метод', y='Кількість опрацьованих зображень', data=data, palette=colors)

# Set the y limit
plt.ylim(0, 184)

# Add labels to the middle of the bars
for index, value in enumerate(values):
    plt.text(x=index, y=value/2, s=f"{value}", color='white', ha="center", fontsize=12)

plt.gca().tick_params(axis='both', which='both', length=0)
plt.gca().spines['top'].set_visible(False)
plt.gca().spines['right'].set_visible(False)
plt.gca().spines['bottom'].set_visible(False)
plt.gca().spines['left'].set_visible(False)

# Show the plot
plt.show()
```
